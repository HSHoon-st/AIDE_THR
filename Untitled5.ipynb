```python
# 행렬식으로 역행렬 계산
def determinant(matrix):
    n = len(matrix)

    # 1x1 행렬
    if n == 1:
        return matrix[0][0]

    # 2x2 행렬
    if n == 2:
        return matrix[0][0] * matrix[1][1] - matrix[0][1] * matrix[1][0]

    # 3x3 이상: 여인수 전개
    det = 0

    for j in range(n):

        minor = []
        for i in range(1, n):
            row = []
            for k in range(n):
                if k != j:
                    row.append(matrix[i][k])
            minor.append(row)

        # 부호
        sign = (-1) ** j

        det += sign * matrix[0][j] * determinant(minor)

    return det


# 행렬의 전치
def transpose(matrix):
    n = len(matrix)

    result = []
    for j in range(n):
        row = []
        for i in range(n):
            row.append(matrix[i][j])
        result.append(row)

    return result


# 행렬식으로 역행렬 계산
def inverse_by_determinant(matrix):
    n = len(matrix)

    det = determinant(matrix)

    if det == 0:
        raise ValueError("역행렬이 존재하지 않습니다. (행렬식 = 0)")

    # 여인수 행렬
    cofactor = []

    for i in range(n):
        row = []

        for j in range(n):
            minor = []

            for r in range(n):
                if r == i:
                    continue

                minor_row = []

                for c in range(n):
                    if c == j:
                        continue

                    minor_row.append(matrix[r][c])

                minor.append(minor_row)

            # 1x1 행렬의 경우
            if n == 1:
                minor_det = 1
            else:
                minor_det = determinant(minor)

            value = ((-1) ** (i + j)) * minor_det
            row.append(value)

        cofactor.append(row)

    # 여인수 행렬의 전치 = 수반행렬
    adjugate = transpose(cofactor)

    # A^(-1) = 1/det(A) * adj(A)
    inverse = []

    for i in range(n):
        row = []

        for j in range(n):
            row.append(adjugate[i][j] / det)

        inverse.append(row)

    return inverse
```

```python
# 가우스-조던 소거법으로 역행렬 계산
def gauss_jordan(matrix):
    n = len(matrix)

    # 원본 행렬을 변경하지 않도록 복사
    a = []

    for i in range(n):
        row = matrix[i][:]

        # 단위행렬 추가
        for j in range(n):
            if i == j:
                row.append(1)
            else:
                row.append(0)

        a.append(row)

    # 가우스-조던 소거
    for i in range(n):

        # 현재 열에서 0이 아닌 피벗 찾기
        pivot = i

        while pivot < n and a[pivot][i] == 0:
            pivot += 1

        # 피벗을 찾지 못하면 역행렬 없음
        if pivot == n:
            raise ValueError("역행렬이 존재하지 않습니다.")

        # 필요한 경우 행 교환
        if pivot != i:
            a[i], a[pivot] = a[pivot], a[i]

        # 피벗을 1로 만들기
        pivot_value = a[i][i]

        for j in range(2 * n):
            a[i][j] /= pivot_value

        # 다른 행의 해당 열을 0으로 만들기
        for k in range(n):
            if k != i:

                factor = a[k][i]

                for j in range(2 * n):
                    a[k][j] -= factor * a[i][j]

    # 오른쪽 부분만 추출
    inverse = []

    for i in range(n):
        row = []

        for j in range(n, 2 * n):
            row.append(a[i][j])

        inverse.append(row)

    return inverse
```

```python
# 행렬 출력
def print_matrix(matrix):
    for row in matrix:
        for value in row:
            print(f"{value:10.4f}", end=" ")
        print()


# 두 행렬 비교
def compare_matrix(A, B):
    n = len(A)

    for i in range(n):
        for j in range(n):
            # 부동소수점 오차 고려
            if abs(A[i][j] - B[i][j]) > 1e-9:
                return False

    return True


#-----------------------------------------------------------------------------
# 실행

# 1. 행렬 크기 입력
n = int(input("행렬 크기 입력: "))

# 2. 행렬 입력
matrix = []

print(f"{n}개의 정수를 공백으로 구분하여 {n}개의 행을 입력하세요.")

for i in range(n):
    while True:
        row = list(map(int, input(f"{i + 1}번째 행: ").split()))

        if len(row) == n:
            matrix.append(row)
            break
        else:
            print(f"{n}개의 정수를 입력해야 합니다.")


# 입력 행렬 출력
print("\n입력한 행렬")
print_matrix(matrix)


# 3. 행렬식으로 역행렬 계산
try:
    inverse1 = inverse_by_determinant(matrix)

    print("\n[행렬식을 이용한 역행렬]")
    print_matrix(inverse1)

except ValueError as e:
    print("\n[행렬식을 이용한 역행렬]")
    print(e)
    inverse1 = None


# 4. 가우스-조던 소거법으로 역행렬 계산
try:
    inverse2 = gauss_jordan(matrix)

    print("\n[가우스-조던 소거법을 이용한 역행렬]")
    print_matrix(inverse2)

except ValueError as e:
    print("\n[가우스-조던 소거법을 이용한 역행렬]")
    print(e)
    inverse2 = None


# 5. 결과 비교
print("\n[결과 비교]")

if inverse1 is not None and inverse2 is not None:
    if compare_matrix(inverse1, inverse2):
        print("역행렬이 동일하다.")
    else:
        print("역행렬이 다름.")
else:
    print("역행렬 존재하지 않음")
```
