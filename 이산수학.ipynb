```python

# 집합 A = {1, 2, 3, 4, 5}
N = 5


# 1. 관계행렬 입력 함수
def input_matrix():
    print("5 x 5 관계행렬을 입력하세요. (0 또는 1)")
    matrix = []

    for i in range(N):
        while True:
            row = list(map(int, input(f"{i+1}행: ").split()))

            if len(row) == N and all(x in (0, 1) for x in row):
                matrix.append(row)
                break

            print("0 또는 1로 이루어진 숫자 5개를 입력하세요.")

    return matrix


# 2. 관계행렬 출력 함수
def print_matrix(matrix):
    for row in matrix:
        print(row)


# 3. 반사 관계 판별 함수
def is_reflexive(matrix):
    for i in range(N):
        if matrix[i][i] != 1:
            return False
    return True


# 4. 대칭 관계 판별 함수
def is_symmetric(matrix):
    for i in range(N):
        for j in range(N):
            if matrix[i][j] != matrix[j][i]:
                return False
    return True


# 5. 추이 관계 판별 함수
def is_transitive(matrix):
    for i in range(N):
        for j in range(N):
            for k in range(N):
                if matrix[i][j] == 1 and matrix[j][k] == 1:
                    if matrix[i][k] == 0:
                        return False
    return True


# 6. 동치관계 판별 및 동치류 출력 함수
def check_equivalence(matrix):
    reflexive = is_reflexive(matrix)
    symmetric = is_symmetric(matrix)
    transitive = is_transitive(matrix)

    print("\n[동치관계 판별]")
    print("반사 관계:", reflexive)
    print("대칭 관계:", symmetric)
    print("추이 관계:", transitive)

    if reflexive and symmetric and transitive:
        print("결과: 동치관계입니다.")
        print_equivalence_classes(matrix)
        return True

    else:
        print("결과: 동치관계가 아닙니다.")
        return False


# 7. 동치류 출력 함수
def print_equivalence_classes(matrix):
    print("\n[각 원소의 동치류]")

    for i in range(N):
        equivalence_class = []

        for j in range(N):
            if matrix[i][j] == 1:
                equivalence_class.append(j + 1)

        print(f"[{i+1}] = {equivalence_class}")


# 8. 반사 페포 함수
def reflexive_closure(matrix):
    closure = [row[:] for row in matrix]

    for i in range(N):
        closure[i][i] = 1

    return closure


# 9. 대칭 페포 함수
def symmetric_closure(matrix):
    closure = [row[:] for row in matrix]

    for i in range(N):
        for j in range(N):
            if matrix[i][j] == 1:
                closure[j][i] = 1

    return closure


# 10. 추이 페포 함수 (워셜 알고리즘)
def transitive_closure(matrix):
    closure = [row[:] for row in matrix]

    for k in range(N):
        for i in range(N):
            for j in range(N):
                if closure[i][k] == 1 and closure[k][j] == 1:
                    closure[i][j] = 1

    return closure


# 11. 페포 적용 및 결과 출력 함수
def process_closure(name, original, closure_function):
    print(f"\n========== {name} ==========")

    print("\n[변환 전 관계행렬]")
    print_matrix(original)

    closure = closure_function(original)

    print("\n[변환 후 관계행렬]")
    print_matrix(closure)

    print(f"\n[{name} 적용 후 동치관계 판별]")
    check_equivalence(closure)


# 12. 메인 함수
def main():
    relation = input_matrix()

    print("\n========== 입력 관계행렬 ==========")
    print_matrix(relation)

    print("\n========== 원래 관계의 판별 ==========")
    result = check_equivalence(relation)

    if not result:
        print("\n========== 페포 변환 시작 ==========")

        process_closure("반사 페포", relation, reflexive_closure)

        process_closure("대칭 페포", relation, symmetric_closure)

        process_closure("추이 페포", relation, transitive_closure)


# 프로그램 실행
if __name__ == "__main__":
    main()
```
